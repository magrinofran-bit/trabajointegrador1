```python
contador = 0
lista_consolidar = []

while contador < 200:
    print('Contador: ',contador)
    params = {'term':'restaurants', 'location':ciudad, 'limit':50, 'offset':contador}
    response = requests.get(api_url, params=params, headers=headers)
    data = response.json()

    df_yelp_v2 = pd.json_normalize(
        data['businesses'],
        record_path=['categories'],
        meta=['id','name','review_count','rating','price'],
        errors='ignore'
    )
    lista_consolidar.append(df_yelp_v2)
    contador += 50
```

```python
df_yelp_vf = pd.concat(lista_consolidar, ignore_index=True)
```
